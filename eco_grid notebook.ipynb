```python
import random
from datetime import datetime

data = []

for i in range(10):
    temp = round(random.uniform(20, 45), 2)
    
    # Solar logic: zyada temperature = zyada solar generation (simple assumption)
    solar = round(random.uniform(0, 3) if temp > 25 else random.uniform(0, 1), 2)

    record = {
        "meter_id": f"MTR_{random.randint(1000, 9999)}",
        "usage_kW": round(random.uniform(0.5, 5.0), 2),
        "voltage": round(random.uniform(210, 240), 2),
        "frequency": round(random.uniform(49.5, 50.5), 2),
        "temperature": temp,
        "solar_kW": solar,
        "latitude": round(random.uniform(22.5, 23.5), 6),
        "longitude": round(random.uniform(75.5, 76.5), 6),
        "timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    }
    data.append(record)

df = spark.createDataFrame(data)
df.show()
```

```python
df_bronze = spark.read.json(bronze_path)

df_bronze.show()
```

silver layer

```python
from pyspark.sql.functions import col, to_timestamp

df_silver = df_bronze.dropna()
```

```python
df_silver = df_silver.withColumn("usage_kW", col("usage_kW").cast("double"))
df_silver = df_silver.withColumn("voltage", col("voltage").cast("double"))
df_silver = df_silver.withColumn("frequency", col("frequency").cast("double"))
```

```python
df_silver = df_silver.withColumn("timestamp", to_timestamp("timestamp"))
```

```python
df_silver.show()
df_silver.printSchema()
```

```python
from pyspark.sql.functions import when

df_theft = df_silver.withColumn(
    "theft_flag",
    when(df_silver["usage_kW"] > 4, "YES").otherwise("NO")
)
```

```python
df_theft.select("meter_id", "usage_kW", "theft_flag").show()
```

```python
from pyspark.sql.functions import when, col

df_maintenance = df_theft.withColumn(
    "maintenance_flag",
    when((col("voltage") < 215) | (col("voltage") > 235), "CHECK")
    .otherwise("NORMAL")
)
```

```python
df_maintenance.select("meter_id", "voltage", "maintenance_flag").show()
```

```python
from pyspark.sql.functions import col, rand, round

df_billing = df_maintenance.withColumn(
    "bill_amount",
    round(col("usage_kW") * 10, 2)
).withColumn(
    "balance",
    round(rand() * 100, 2)
)

```

```python
from pyspark.sql.functions import when

df_billing = df_billing.withColumn(
    "status",
    when(col("balance") < col("bill_amount"), "DISCONNECTED")
    .otherwise("ACTIVE")
)
```

```python
df_billing.select("meter_id", "usage_kW", "bill_amount", "balance", "status").show()
```

```python
from pyspark.sql.functions import when, col

df_eco = df_billing.withColumn(
    "eco_score",
    when((col("usage_kW") < 2) & (col("solar_kW") > 1), "GREEN")
    .when(col("usage_kW") > 4, "RED")
    .otherwise("YELLOW")
)
```

```python
df_eco = df_eco.withColumn(
    "reward_points",
    when(col("eco_score") == "GREEN", 50)
    .when(col("eco_score") == "YELLOW", 20)
    .otherwise(0)
)
```

```python
df_eco.select("meter_id", "usage_kW", "solar_kW", "eco_score", "reward_points").show()
```

gold layer (aggregation)

```python
from pyspark.sql.functions import sum, round

df_gold = df_eco.groupBy("latitude", "longitude").agg(
    round(sum("usage_kW"), 2).alias("total_usage"),
    round(sum("bill_amount"), 2).alias("total_revenue")
)
```
